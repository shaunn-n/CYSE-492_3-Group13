# Power Analysis Instructions
This documentation will briefly explain how the following scripts work:
- `optimal_poweranalysis_jtop.py`
- `VLM_LiveVideo_Prompting_Latency.py`

and how to use them to perform power analysis

#### This portion of the jupyter notebook contains complete documentation on how power analysis is completed

### Overview:
Performing power analysis involves three tasks in the following order:
1) Run one out of the 5 VLMs via Ollama
2) Run the Python script that will continuously prompt the VLM and take in/analyze a live camera feed
3) Begin measuring GPU metrics (for power analysis) by running a separate script


## Setting up a VLM
You can view all of the VLMs installed in the terminal using the following command:
```bash
ollama list
```
There you'll see 5 of the following VLMs:
- Moondream
- aeline/opan
- Qwen3-vl 
- Minicpm-v4.6
- SmolVLM

Use the exact model names from `ollama list`. For the sake of simplicity, we will use Moondream as an example.
```bash
ollama run moondream:latest
```

From there, you will see a loading symbol, indicating that the VLM you've chosen it booting up. You can test to see if it's working by entering a prompt such as "Hello, how are you?"

# Setting up the live camera feed and continous prompting
This is where we will begin to use the VLM_LiveVideo_Prompting_Latency.py script

Open a new terminal window or tab, and change your directory to ensure you're in the same location as the prompting script.

For instance, if the script was saved in the Documents folder, you would use the following command:
```bash
cd /Documents
```

For this particular script, when you are trying to run it, you are required to enter the exact name of the VLM you are running, and the file name ending in .csv, all in one line in the terminal. Ensure that you type `python3 VLM_LiveVideo_Prompting_Latency.py (insert VLM) (insert file name).csv`

**FYI:** The script also measures latency which is why it also requires a .csv file name input from the user. For simplicity, we decided to combine latency measurement and continuous prompting + live camera feed rather than have two separate scripts

For example:
```bash 
python3 VLM_LiveVideo_Prompting_Latency.py moondream:latest example.csv
```

An error will occur if you only try to run the script without enter any of the inputs


**When you are finished:**

Use Ctrl+C in the terminal to stop the script. Note that the script will run forever until the user inputs Ctrl+C.

# Recording GPU Metrics
Here is where we'll use `optimal_poweranalysis_jtop.py` to actually track power consumption over a certain period of time

Again, change directories to where the script is located.

Then, run the script. It will prompt some inputs regarding the file name of the .csv, the sampling rate (interval of time between measurements), and the duration of time at which the script will capture data. It should look something like this in the terminal:

```bash
python3 optimal_poweranalysis_jtop.py
Enter a file name (end it with .csv): 
Moondream_poweranalysis_MAXN_noattack_TodaysDate.csv

Enter a time interval to adjust the sampling rate (e.g. entering 10 means measurements will be taken every 10 seconds):
1

Enter duration to log data:
20

Beginning logging process for 20 minutes
```

After the set duration (in this case 20 minutes), the script will say `Logging period complete. Check folder for .csv file`. The .csv file will be saved in the same folder where the script is located.

If for whatever reason you need to stop the script before the duration of time has passed, use Ctrl+C. 

**Important things to note:**
- The name of the csv file should be as specific as possible while also being concise. In this case we followed this naming convention: *ModelName_poweranalysis_PowerLimitSetting_AttackUsedIfNecessary_TodaysDate.csv*
- The script takes float values for both the time interval and duration inputs. This is so we are able to increase the sampling rate, allowing for more precise measurement to detect any triggers/spikes in power.
- Stopping the script with Ctrl+C will not save any recorded data. You must start over and follow through with your desired duration of time in order for the .csv file to save captured data.
